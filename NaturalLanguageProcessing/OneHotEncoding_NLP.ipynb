## One Hot Encoding for NLP

One-Hot Encoding (OHE) is a process of converting categorical data (like words in NLP) into a numerical format that machine learning algorithms can understand. In OHE, each unique word in a vocabulary is represented as a binary vector where all elements are 0 except for one, which is 1, indicating the presence of that specific word. The position of the '1' corresponds to the unique index assigned to that word. This method helps to avoid implying any ordinal relationship between words.

* **Simple Example:**

```python
# Sample vocabulary
vocabulary = ['apple', 'banana', 'orange', 'grape']

# Create a mapping from word to index
word_to_index = {word: i for i, word in enumerate(vocabulary)}

def one_hot_encode(word, vocab_size):
    vector = np.zeros(vocab_size)
    if word in word_to_index:
        vector[word_to_index[word]] = 1
    return vector

# Encode some words
vocab_size = len(vocabulary)
encoded_apple = one_hot_encode('apple', vocab_size)
encoded_banana = one_hot_encode('banana', vocab_size)
encoded_unknown = one_hot_encode('kiwi', vocab_size) # OOV word

print(f"'apple' encoded: {encoded_apple}")
print(f"'banana' encoded: {encoded_banana}")
print(f"'kiwi' (unknown) encoded: {encoded_unknown}")
```

### Advantages and Disadvantages of One Hot Encoding

| Advantages                            | Disadvantages                                     |
|---------------------------------------|---------------------------------------------------|
| Simple to understand and implement    | High dimensionality (curse of dimensionality)     |
| Ensures orthogonality of words        | Does not capture semantic similarity between words|
| Works well with simple models         | Leads to sparse matrices, inefficient for large vocabularies|
| No assumption about word relationships| Out-of-vocabulary (OOV) words cannot be handled directly|

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
corpus = [
    "Artificial intelligence is transforming the world.",
    "Natural language processing enables machines to understand text.",
    "Machine learning models require large datasets for training.",
    "Deep learning uses neural networks to solve complex problems.",
    "Computer vision allows computers to interpret visual information.",
    "Robotics combines engineering and computer science fields.",
    "Data science involves extracting insights from structured data.",
    "Algorithms are the core of modern computing systems.",
    "Big data technologies handle massive amounts of information.",
    "Cloud computing provides scalable resources for applications."
]
```

```python
corpus[0].lower()
```

```python
## Tokenizing the words in corpus

#Creating a Vocabulary
words = set()
for sentences in corpus:
    for word in sentences.lower().replace('.','').split():
        words.add(word) 

# sorting the vocabulary
vocab = sorted(list(words))
word_to_index = {word: i for i, word in enumerate(vocab)}

## defining a function to convert text to one hot encoding
def convert_to_one_hot(sentence):
    vocab_size = len(vocab)
    vector = np.zeros((vocab_size), dtype="int16")
    for word in sentence.lower().replace('.','').split():
        vector[word_to_index[word]] = 1 # finding the index of the word and setting it to 1
    return vector

## Example for first sentence
first_sentence_vector = convert_to_one_hot(corpus[0])
print("First Sentence: ",corpus[0])
print("First Sentence Vector: ",first_sentence_vector)


# Converting the corpus to OneHotEncoded Vector
OHE_DF = pd.DataFrame([convert_to_one_hot(sentence) for sentence in corpus], columns = vocab)
print("\n\nOne Hot Encoded DataFrame: ")
OHE_DF
```
