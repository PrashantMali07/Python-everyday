```python
import pandas as pd
import numpy as np
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
## Creating object of CountVectorizer, PorterStemmer
ps=PorterStemmer()
lemma = WordNetLemmatizer()
vectorizer = CountVectorizer(max_features=150)
```

```python
corpus = [
    "Artificial intelligence is transforming the world.",
    "Natural language processing enables machines to understand text.",
    "Machine learning models require large datasets for training.",
    "Deep learning uses neural networks to solve complex problems.",
    "Computer vision allows computers to interpret visual information.",
    "Robotics combines engineering and computer science fields.",
    "Data science involves extracting insights from structured data.",
    "Algorithms are the core of modern computing systems.",
    "Big data technologies handle massive amounts of information.",
    "Cloud computing provides scalable resources for applications."
]
```

## Using PorterStemmer

```python
ps_corpus=[]
stop_words = set(stopwords.words("english"))

for sent in corpus:
    sentence = sent.lower().replace(".", "")
    sentence = sentence.split()
    sentence = [ps.stem(word) for word in sentence if word not in stop_words]
    sentence = " ".join(sentence)

    ps_corpus.append(sentence)

print(ps_corpus)
```

```python
## Creating Vector
ps_vec = vectorizer.fit_transform(ps_corpus).toarray()

for idx in range(5): print(f"Sentence: {ps_corpus[idx]}\nArray: {ps_vec[idx]}"); print('-'*50)

## creating a Dataframe
words=set()
for sentence in ps_corpus:
    for word in sentence.split():
        words.add(word)

vocab=list(words)

ps_df = pd.DataFrame(ps_vec, columns=vocab)
print(f"\nShape of the Dataframe: {ps_df.shape}\n")
ps_df
```

## Using WordNetLemmatizer

```python
lemma_corpus=[]

for sent in corpus:
    sentence = sent.lower().replace(".", "")
    sentence = sentence.split()
    sentence = [lemma.lemmatize(word) for word in sentence if word not in stop_words]
    sentence = " ".join(sentence)

    lemma_corpus.append(sentence)

print(lemma_corpus)
```

```python
## Creating Vector
lemma_vec = vectorizer.fit_transform(lemma_corpus).toarray()

for idx in range(5): print(f"Sentence: {lemma_corpus[idx]}\nArray: {lemma_vec[idx]}\n"); print('-'*50)

## creating a Dataframe
words=set()
for sentence in lemma_corpus:
    for word in sentence.split():
        words.add(word)

vocab=list(words)

lemma_df = pd.DataFrame(lemma_vec, columns=vocab)
print(f"\nShape of the Dataframe: {lemma_df.shape}\n")
lemma_df
```
