```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([FAQS])
```

```python
print(f"Total number of words in the vocabulary: {len(tokenizer.word_index)}")
```

```python
input_sequences = []
for line in FAQS.split('\n'):
    ## Storing the array of each sentences
    sentence_arr = tokenizer.texts_to_sequences([line])[0]
    ## Generating n-grams from each sentence
    for idx in range(1,len(sentence_arr)):
        input_sequences.append(sentence_arr[:idx + 1])
        print(sentence_arr[:idx + 1])
```

```python
## Need to apply the pre '0' padding to make all the sequences of same length
## Required the max length array in input_sequences
max_len = max([len(arr) for arr in input_sequences])
print(f"Maximum length of the sequence is: {max_len}")
```

```python
padded_input_sequences = pad_sequences(input_sequences, maxlen=max_len, padding='pre',dtype='int16')
print("Type of the padded array: " + str(type(padded_input_sequences)))
print("Shape of the padded array: " + str(padded_input_sequences.shape))
padded_input_sequences
```

```python
## Splitting the data into input and output
## Last index token in 'padded_input_sequences' will be the output token

X, y = padded_input_sequences[:,:-1], padded_input_sequences[:,-1]
print("Shape of input data: " + str(X.shape))
print("Shape of output data: " + str(y.shape))
```

```python
## Considering it to be a MULTI CLASS CLASSIFICATION problem for next word prediction
y = to_categorical(y, num_classes=len(tokenizer.word_index) + 1)
y.shape
```

```python
## Output array 'y' contains the one-hot encoded vectors for each output token:
y
```

```python
model = Sequential()
model.add(Embedding(input_dim=len(tokenizer.word_index) + 1, output_dim=100, input_length=max_len - 1))
model.add(LSTM(156))
model.add(Dense(len(tokenizer.word_index) + 1, activation='softmax'))
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
model.summary()
```

```python
model.fit(X, y, epochs=100, verbose=1,validation_split=0.1)
```

```python
## Predicting the next word given a sequence

import time
text = "These ads are six"
text = text.lower()

for i in range(7):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=max_len - 1, padding='pre')
  # predict
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
      time.sleep(0.5)
```
