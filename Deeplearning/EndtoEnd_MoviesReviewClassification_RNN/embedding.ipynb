```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences = [
    "The quick brown fox jumps over the lazy dog."
    "I love playing football with my friends."
    "The sun rises in the east and sets in the west."
    "Machine learning is a subset of artificial intelligence."
    "The cat purrs contentedly on my lap."
    "I am excited to try out this new restaurant tonight."
    "The capital of France is Paris."
    "The new iPhone is a game-changer."
    "I am learning to play the guitar in my free time."
    "The weather is beautiful today, isn't it?"
]

sentences_ = [sentence.lower() for sentence in sentences]
sentences_ = [sentence.split('.') for sentence in sentences_]
sentences_ = sentences_[0]
sentences_
```

```python
## Defining vocabulary size
vocab_size = 10000

## One hot representation
one_hot_mat = [one_hot(words, vocab_size) for words in sentences_]
one_hot_mat
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
## Finding the max length of the one hot encoded vectors
sent_len = max(len(i) for i in one_hot_mat)+1
print(f"Maximum length of the sentences: {sent_len}")
```

```python
## Creating embedded vectors with padding
embedded_docs = pad_sequences(one_hot_mat,padding='pre',maxlen=sent_len)

print("Checking vector after padding:\n", embedded_docs)
```

```python
## Defining dimention to create embedding vector
dim = 10

## Creating the model
model = Sequential()
model.add(Embedding(vocab_size, dim, input_length=sent_len))
model.compile('adam', 'mse')
model.summary()
```

```python
predicted_arr = model.predict(embedded_docs)
display(predicted_arr)
```
