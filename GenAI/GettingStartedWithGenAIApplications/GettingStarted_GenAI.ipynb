# Getting started with LangChain & OpenAI

#### In this Quickstart we'll see how to;

* Get setup with LangChain, LangSmith and LangServe
* Use the most basic and common components of LangChain: prompt templates, models, output parsers.
* Build a simple application with LangChain
* Trace the Application with LangSmith
* Serve your application with LangServe

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
```

#### Testing OpenAI API Key

```python
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

## use 'gpt-4o-mini' for faster response and lower cost, but 'gpt-5.2' for better quality
response = client.responses.create(
    model="gpt-5.2",
    input="Write a short bedtime story about a unicorn."
)

print(response.output_text)
```

```python
## Loading .env

os.environ['OPENAI_API_KEY']=os.getenv("OPENAI_API_KEY")
os.environ['LANGCHAIN_API_KEY']=os.getenv("LANGCHAIN_API_KEY")
os.environ['LANGCHAIN_TRACING_V2']="true"
os.environ['LANGCHAIN_PROJECT']=os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini")
print(llm)
```

```python
## Inpu and get response from OpenAI LLM
response = llm.invoke("What is LangChain?")
print(response)
print('-'*50)
print(response.content)
```

#### Chat Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are an expert AI Engineer, provide me answer based on the question"),
        ("user","{input}")
    ]
)
```

```python
chain=prompt|llm ## chain is a combination of prompt and llm, we can use it to get response from llm based on the prompt

response=chain.invoke({"input":"Can you tell me about Bagging & Boosting in Machine Learning?"})
print(response)
```

```python
## Gettintg response content
print(response.content)
```

#### String Output Parser

```python
from langchain_core.output_parsers import StrOutputParser

output_parser=StrOutputParser()
chain_with_parser=prompt|llm|output_parser

response=chain_with_parser.invoke({"input":"What is CrewAI & AgenticAI?"})
```
