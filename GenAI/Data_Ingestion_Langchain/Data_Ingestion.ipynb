## Data Ingestion using LangChain-Community

#### 1. Text Loader

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("/home/prashant/Documents/Remove and Reinstall NVIDIA.txt")
loader
```

```python
## Checking the content of the loaded document
documents = loader.load()
print(documents)
```

#### 2. PDF Loader

```python
from langchain_community.document_loaders import PyPDFLoader

pdf_loader = PyPDFLoader("/home/prashant/Documents/2306.07303v1.pdf")
documents_pdf = pdf_loader.load()
documents_pdf
```

#### 3. WebBaseLoader

```python
## Web Base Loader
from langchain_community.document_loaders import WebBaseLoader
import bs4

web_loader = WebBaseLoader(web_paths=("https://lilianweng.github.io/posts/2023-06-23-agent/",),
                           bs_kwargs=dict(parse_only=bs4.SoupStrainer(class_=("post-title","post-content","post-header"))
                            ))
web_loader
```

```python
web_loader.load()
```

#### 4. Arxiv Loader

```python
## Arxiv Loader
from langchain_community.document_loaders import ArxivLoader

loader = ArxivLoader(query="1605.08386", load_max_docs=2).load()
len(loader)
```
