```python
from langchain_community.document_loaders import PyPDFLoader

pdf_loader = PyPDFLoader("/home/prashant/Documents/2306.07303v1.pdf")
documents_pdf = pdf_loader.load()
documents_pdf
```

#### How to recursively split text by characters

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
## Define the text splitter object
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks = text_splitter.split_documents(documents_pdf)
chunks
```

```python
## Checking the number of chunks
len(chunks)
```

```python
## Checking first chunk
print(chunks[0].metadata)
print(chunks[0].page_content)
```

#### How to convert Text into a Document type

```python
with open("/home/prashant/Documents/Remove and Reinstall NVIDIA.txt", "r") as f:
    text = f.read()
print(text)
```
