```python
from langchain_ollama import ChatOllama
```

```python
## Calling the model
llm = ChatOllama(
    model='llama3.2-vision:latest',
    temperature=0
)

response = llm.invoke("Explain the concept of quantum entanglement in one sentence.")
print(response)
```

```python
print(response.content)
```

```python
## For getting the answers directly

from langchain_ollama import OllamaLLM

llm = OllamaLLM(model='llama3.2-vision:latest')
response = llm.invoke("The Prime Minister of North Korea is")
print(response)
```

```python
## Calling an embedding models and checking the results

from langchain_ollama import OllamaEmbeddings

embedding_model = OllamaEmbeddings(model='embeddinggemma:latest')

# Embed a single query
embedded_result = embedding_model.embed_query("What is Time Space Complexity?")
print(embedded_result)
```

```python
## Embedding documents file

doc_result = embedding_model.embed_documents(["/home/prashant/Documents/can we built a chat window kafka.txt", "/home/prashant/Documents/Data Analysis Roadmap.txt"])
print("Embedding length:", len(doc_result))
```
