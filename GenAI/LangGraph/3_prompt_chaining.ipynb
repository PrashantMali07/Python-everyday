## Prompt Chaining

### Topic >> LLM to Generate detailed outline for Topic >> Generate Blog with Topic + Detailed Outline >> Blog

#### Workflow (2): START >> generate_outline >> generate_blog >> END

```python
import os
from dotenv import load_dotenv

from langgraph.graph import StateGraph, START, END
from langchain_ollama import ChatOllama
from typing import TypedDict

load_dotenv()
```

```python
## Initialize LLM
llm = ChatOllama(model='gemma4:31b-cloud')
```

```python
class BlogState(TypedDict):

    title: str
    outline: str
    content: str
    rating: int
```

```python
def generate_outline(state: BlogState)-> BlogState:

    title = state['title']

    prompt = f"Generate a detailed outline for the topic: {title}"

    detailed_outline = llm.invoke(prompt).content

    state['outline'] = detailed_outline

    return state

def generate_blog(state: BlogState)-> BlogState:

    title = state['title']
    outline = state['outline']

    prompt = f"Create a detailed blog on the title: \n{title} \n\nusing the given outline:\n{outline}"

    response = llm.invoke(prompt).content

    state['content'] = response

    return state

def evaluate_(state: BlogState)->BlogState:

    outline = state['outline']
    content = state['content']

    prompt = f"Based on the outline: \n{outline} and \n\ncontent:\n{content}, \nEvaluate and generate a score between 1 to 5."

    score = llm.invoke(prompt).content

    state['rating'] = score

    return state
```

```python
# Define Graph
graph = StateGraph(BlogState)

# add node
graph.add_node('generate_outline', generate_outline)
graph.add_node('generate_blog', generate_blog)
graph.add_node('evaluate', evaluate_)

# add edges
graph.add_edge(START, 'generate_outline')
graph.add_edge('generate_outline','generate_blog')
graph.add_edge('generate_blog','evaluate')
graph.add_edge('evaluate',END)

# compile
workflow = graph.compile()
```

```python
initial_state = {'title': 'Rise of AI in the World'}

final_state = workflow.invoke(initial_state)
```
