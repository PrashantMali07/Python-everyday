```python
from langgraph.store.memory import InMemoryStore
```

```python
#creating a store
store = InMemoryStore()
```

```python
# creating a namespace
namespace = ('user','u1')
```

### Creating Memories

#### 1. Adding Memories

```python
store.put(
    namespace=namespace,
    key='1',
    value={'data': 'user likes cheese'}
)

store.put(
    namespace=namespace,
    key='2',
    value={'data': 'user prefers train inplace of bus'}
)
```

```python
# creating another namespace
namespace = ('user','u2')

# add memories
store.put(
    namespace=namespace,
    key='1',
    value={'data': 'user likes pasta'}
)

store.put(
    namespace=namespace,
    key='2',
    value={'data': 'user prefers using long knife'}
)
```

### 2. Retrieving Memories

```python
# store.get(namespace,key)

namespace = ('user','u1')

store.get(namespace,'2')
```

### 3. Retrieving All Memories

```python
items = store.search(namespace)

for item in items:
    print(item)
```

### 4. Semantic Search using serch() method

```python
from langchain_ollama import OllamaEmbeddings

# initializing embedding model
embed = OllamaEmbeddings(model='qwen3-embedding:8b',dimensions=2048)

# creating Memory Store with embedding model for Semantic Search capabilities
store = InMemoryStore(index={'embed': embed, 'dims':2048})
```

```python
namespace = ('user','u1')

store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
items = store.search(namespace, query="what is the user currently learning", limit=1)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace, query="what are user's preferences", limit=4)

for item in items:
    print(item.value)
```
