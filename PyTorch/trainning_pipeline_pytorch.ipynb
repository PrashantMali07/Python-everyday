```python
# Splitting Dataset
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:], df.iloc[:,0], test_size=0.2, random_state=42)

print("Shape of the splitted dataset:", [X_train.shape, X_test.shape, y_train.shape, y_test.shape])
```

```python
## Scaling the dataset
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Converting the numpy array to pytorch tensor

X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

## Defining the ANN Model using PyTorch

```python
class SimpleNN():
  ## Defining Constructor
  def __init__(self, X):

    self.weights = torch.rand(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias = torch.zeros(1,dtype=torch.float64,requires_grad=True)

  # Forward training function
  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_hat = torch.sigmoid(z)
    return y_hat

  # Function for calculating Loss
  def loss_function(self, y_hat, y):
    # Clamp prediction to avoid log(0)
    epsilon = 1e-7
    y_hat = torch.clamp(y_hat, epsilon, 1-epsilon)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_hat) + (1-y_train_tensor) * torch.log(1-y_hat)).mean()
    return loss
```

## Defining trainning parameters & model

```python
learning_rate = 0.1
epochs = 50

## Create model
model = SimpleNN(X_train_tensor)

## Define loop
for epoch in range(epochs):

  # Forward pass
  y_pred = model.forward(X_train_tensor)

  # Calculate loss
  loss = model.loss_function(y_pred, y_train_tensor)

  # Backward pass
  loss.backward()

  # Update Parameters
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # Zero Gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

## Evaluation

```python
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.5).float()
  print(f"Predicted Array:\n{y_pred}")
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
