## Mathematical Operation

#### 1. Scalar Operations

```python
# Creating a tensor
x = torch.rand(2,3)
x
```

```python
# addition
print(x + 2)

# substraction
print(x - 2)

# multiplication
print(x * 2)

# division
print(x / 3)

# int division
print((x * 100)//3)

# mod
print(((x * 100)//3)%2)

# power
print(x ** 2)
```

#### 8. Copying a Tensor

```python
ct_a = torch.rand(2,3)
ct_b = ct_a
```

```python
ct_a
```

```python
## If we make changes ct_a it will reflect it in ct_b too, as it is a assignment operator

## Checking if any changes done in ct_a will reflect in ct_b

ct_a[0][0] = 0

ct_a
```

Changes made inside ct_b variable also.

```python
ct_b
```

```python
print(f"Memory Location ID, ct_a: {id(ct_a)}")
print(f"Memory Location ID, ct_b: {id(ct_b)}")
```

Which is same.

```python
## But if we use clone function it will create a copy with new memory location

ct_b = ct_a.clone()

print("Now again checking the Memory location ID.")
print(f"Memory Location ID, ct_a: {id(ct_a)}")
print(f"Memory Location ID, ct_b: {id(ct_b)}")
```

Which is different now, as the `clone()` function create a copy of other variable and If we make any changes inside a primary variable it will not reflect in secondary variable.

#### 9.Tensor Operation on GPU

```python
## Checking for the available GPU

torch.cuda.is_available()
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
device
```

```python
## Creating a tensor inside a GPU

torch.rand(5,4, device=device)
```

```python
## Moving an existing tensor to GPU

new_tensor = torch.rand(5,4)
gpu_tensor = new_tensor.to(device) ## moving tensor to GPU with a new variable

print(gpu_tensor)
```

```python
## Defining a matrix multiplication and com-paring time of execution with GPU & CPU

import time

## defining a size of matrix
size = 10000 ## For large size perormance comparision

## Create a random matrices on CPU
mat_cpu1 = torch.randn(size,size)
mat_cpu2 = torch.randn(size,size)

#Measure time on CPU
start_time  = time.time()
mat_result_cpu = torch.matmul(mat_cpu1,mat_cpu1) ## Matrix multiplication on CPU
cpu_time = time.time() - start_time

print(f"Time of CPU: {cpu_time:.4f} seconds")


## Moving matrices to GPU\
mat_gpu1 = mat_cpu1.to(device)
mat_gpu2 = mat_cpu2.to(device)

#Measure time on CPU
start_time  = time.time()
mat_result_gpu = torch.matmul(mat_gpu1,mat_gpu2) ## Matrix multiplication on GPU
torch.cuda.synchronize() # Ensure all GPU operations are complete
gpu_time = time.time() - start_time

print(f"Time of GPU: {gpu_time:.4f} seconds")
print(f"By {((cpu_time-gpu_time)/cpu_time)*100:.2f}% GPU is faster then CPU.")
```
