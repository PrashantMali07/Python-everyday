# 1.Overview on PySpark

Here are the steps to install a Python environment using **conda** for setting up **PySpark**:

***

## Step 1: Install Conda (if not installed)
- Download and install Miniconda or Anaconda from their official websites.

***

## Step 2: Create a New Conda Environment

Run this command to create a new environment (e.g., named `pyspark-env`) with Python installed:

```bash
conda create -n pyspark-env python=<version> -y
```

***

## Step 3: Activate the Environment

Activate the environment using:

```bash
conda activate pyspark-env
```

***

## Step 4: Install PySpark

Install PySpark within this environment:

```bash
conda install -c conda-forge pyspark -y
```

```bash
pip install pyspark -y
```

This installs PySpark along with its dependencies, including Java if needed.

***

## Step 5: Verify PySpark Installation

Open Python interpreter or a script and run:

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("TestApp").getOrCreate()
print(spark.version)
```

If no errors and Spark version prints, PySpark is set up correctly.

***

## Optional: Install Jupyter Notebook (to use notebooks)

```bash
conda install jupyter -y
```

Then launch with:

```bash
jupyter notebook
```

***

### 4. Handling the missing value in DataFrame using PySpark

- Dropping columns
- Dropping Rows
- Various parameters in dropping Functionalities
- Handling Missing values by `Mean` `Median` and `Mode`

````python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Dataframe").getOrCreate()
````

````python
df_spark=spark.read.csv("employee.csv", header=True, inferSchema=True)
df_spark.show()
````

````python
## 1.Drop the columns
df_spark.drop('Name').show()
````

````python
## 2. Dropping the rows with null values
df_spark.na.drop().show()  # Drops rows with any null values
````

````python
## 3. Various parameters of na.drop() functionality

## understanding 'how' 'thresh' 'subset' parameters of na.drop() function

## how = 'any' or 'all' (default is 'any')

### any --> Stands for if any null value is present in a row, drop that row

df_spark.na.drop(how='any').show()  # This is default behaviour of na.drop() function
````

````python
### all --> Stands for if all values are null in a row, drop that row

df_spark.na.drop(how='all').show()
````

````python
### thresh = int (default is None) --> Stands for require that given number of non-null values to keep the row

df_spark.na.drop(how='any',thresh=3).show()  # Keep rows with at least 4 non-null values
````

````python
## subset = list of column names (default is None) --> Stands for consider only these columns for null values

df_spark.na.drop(how='any',subset=['Name']).show()  # will drop rows where 'Name' column has null values
````

````python
## Fill the missing values

df_spark.na.fill("No Name").show()  # Fill all null values with "No Name" only in string type columns
````

````python
df_spark.na.fill(0).show()  # Fill all null values with 0 only column having numeric type
````

````python
## Handling missing values in PySpark DataFrame with Mean, Median and Mode

### Importing and creating imputer functionality
from pyspark.ml.feature import Imputer

imputer=Imputer(inputCols=['Age','Salary'],outputCols=['{}_imputed'.format(c) for c in ['Age','Salary']]).setStrategy('mean')
````

````python
imputer.fit(df_spark).transform(df_spark).show()  # Filling missing values with mean of the column
````

````python
# Stop the Spark session
spark.stop()
````

##  5. Filter operation with Dataframe using PySpark

- Filter Operation
- AND `&`
- OR `|`
- EQUAL TO `==`
- Inverse or NOT `~`

````python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Dataframe").getOrCreate()
````

````python
df_filter = spark.read.csv("iris.csv", header=True, inferSchema=True)
df_filter.show()
````

````python
## Filter

### 1. Filter 'sepal_length' greater than or equal to 5.0
df_filter.filter('sepal_length >= 5.0').show()
````

````python
## getting unique values of 'variety' column

df_filter.select('variety').distinct().show()
````

````python
### 2. Filter dataframe where 'variety' is 'Versicolor' and show only 'sepal_length' and 'petal_length' columns using EQUALS TO '==' operator
df_filter.filter("variety == 'Versicolor'").select(['sepal_length', 'petal_length']).show()
````

````python
### loading 'employee.csv' file to filter

df_emp = spark.read.csv("employee.csv", header=True, inferSchema=True)
df_emp.show()
````

````python
## 3. filter employees name where the salary is less than or equal to 50000 and dropping the null values

df_emp.na.drop().filter("Salary <= 50000").select('Name').show()
````

````python
## 4. filter the employee name and department where the salary is above 50000 and age is between 25 and 30, using AND '&' operator
df_emp.na.drop().filter((df_emp['Salary'] > 50000) & (df_emp['Age'].between(25,30))).select(['Name','Department']).show()
````

````python
## 5. filter the employee name and department where the salary is above 50000 and age is between 25 and 30, using OR '|' operator
df_emp.na.drop().filter((df_emp['Salary'] > 50000) | (df_emp['Age'].between(25,30))).select(['Name','Department']).show()
````

````python
## 6. filter the employee name and department where the salary is above 50000 and age is between 25 and 30, using NOT or INVERSE '~' operator
df_emp.na.drop().filter(~(df_emp['Salary'] > 50000)).select(['Name','Department','Salary']).show() # This will show employees with salary less than or equal to 50000
````

````python
# Stop the Spark session
spark.stop()
````

## 6. Working with Aggregate and GroupBy function using PySpark

````python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Dataframe").getOrCreate()
````

````python
df_emp = spark.read.csv("employee_dup.csv", header=True, inferSchema=True)
df_emp.show()
````

````python
df_emp.printSchema()
````

````python
## 1. GroupBy with sum()

df_emp.groupBy('Name').sum().show()  # sum of employees salary for each employee name
````

````python
## 2. GroupBy with sum() and sorting in ascending order

df_emp.groupBy('Name').sum().sort(2,ascending=True).show()  # sum of employees salary for each employee name, in sort() '2' stands for 2nd column which is 'sum(Salary)'

## 2. GroupBy with sum() and sorting in descending order

# df_emp.groupBy('Name').sum().sort(2,ascending=False).show()
````

````python
## Groupby with sum() and checking the sum of salary in department
df_emp.groupBy('Department').sum().sort(2).show()  # sum of employees salary for each department
````

**Observation:** Finance department getting Higher Salary

````python
## Groupby with mean()

df_emp.groupBy('Department').mean().show()
````

````python
## Groupby with count()

df_emp.groupBy('Department').count().show()
````

````python
## Groupby with max()

df_emp.groupBy('Department').max().show()
````

````python
## Groupby with min()

df_emp.groupBy('Department').min().show()
````

````python
## Aggregate functions

df_emp.agg({'Salary':'sum'}).show()  # total sum of salary of all employees
````

````python
df_emp.agg({'Salary':'max'}).show()  # Maximum of salary of all employees
````

````python
df_emp.agg({'Salary':'min'}).show()  # Minimum of salary of all employees
````

````python
# Stop the Spark session
spark.stop()
````
